```python

import pandas as pd
from pathlib import Path

# Locate the project and task directories
current_dir = Path.cwd()

if current_dir.name == "Task-1-Data-Exploration":
    project_dir = current_dir.parent
    task1_dir = current_dir
else:
    project_dir = current_dir
    task1_dir = project_dir / "Task-1-Data-Exploration"

data_path = project_dir / "data" / "Railway_info.csv"
task1_dir.mkdir(parents=True, exist_ok=True)

# Load the original dataset
df = pd.read_csv(data_path)

print("=" * 50)
print("TASK 1.1: LOAD AND INSPECT DATA")
print("=" * 50)

print("\n1. First 10 rows:")
display(df.head(10))

print("\n2. Dataset dimensions:")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

print("\n3. Column names:")
print(df.columns.tolist())

print("\n4. Data types and non-null counts:")
df.info()

print("\n5. Missing values per column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\n6. Exact duplicate rows:")
print(df.duplicated().sum())
```

```python

print("=" * 50)
print("TASK 1.2: BASIC STATISTICS")
print("=" * 50)

# Total records and unique train numbers
print(f"\nTotal train records: {len(df)}")
print(f"Unique train numbers: {df['Train_No'].nunique()}")

# Unique stations
print(
    "Unique source stations:",
    df["Source_Station_Name"].nunique()
)

print(
    "Unique destination stations:",
    df["Destination_Station_Name"].nunique()
)

# Most common source stations
print("\nTop 10 most common source stations:")
display(
    df["Source_Station_Name"]
    .value_counts()
    .head(10)
    .to_frame("Train Count")
)

# Most common destination stations
print("\nTop 10 most common destination stations:")
display(
    df["Destination_Station_Name"]
    .value_counts()
    .head(10)
    .to_frame("Train Count")
)

# Numerical summary
print("\nNumerical summary:")
display(df[["Train_No"]].describe())
```

```python

print("=" * 50)
print("TASK 1.3: DATA CLEANING")
print("=" * 50)

# Create a copy so the original dataset remains unchanged
df_clean = df.copy()

# Check missing values before cleaning
print("\n1. Missing values before cleaning:")
display(df_clean.isnull().sum().to_frame("Missing Values"))

# Remove rows missing essential information
required_columns = [
    "Train_No",
    "Train_Name",
    "Source_Station_Name",
    "Destination_Station_Name",
    "days"
]

rows_before = len(df_clean)

df_clean = df_clean.dropna(
    subset=required_columns
).copy()

print(
    "\nRows removed due to missing essential values:",
    rows_before - len(df_clean)
)

# Standardize text whitespace and capitalization
text_columns = [
    "Train_Name",
    "Source_Station_Name",
    "Destination_Station_Name"
]

for column in text_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
        .str.upper()
    )

# Standardize weekday values
df_clean["days"] = (
    df_clean["days"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Correct known weekday variants
day_mapping = {
    "mondayd": "monday",
    "tuesdayd": "tuesday",
    "wednesdayd": "wednesday",
    "thursdayd": "thursday",
    "fridayd": "friday",
    "saturdayd": "saturday",
    "sundayd": "sunday"
}

df_clean["days"] = df_clean["days"].replace(day_mapping)
df_clean["days"] = df_clean["days"].str.capitalize()

# Verify weekday values
print("\n2. Unique operating days:")
print(sorted(df_clean["days"].dropna().unique()))

# Check duplicate rows after cleaning
print("\n3. Exact duplicate rows after cleaning:")
print(df_clean.duplicated().sum())

# Check missing values after cleaning
print("\n4. Missing values after cleaning:")
display(df_clean.isnull().sum().to_frame("Missing Values"))

# Display cleaned dataset information
print("\n5. Cleaned dataset dimensions:")
print(f"Rows: {df_clean.shape[0]}")
print(f"Columns: {df_clean.shape[1]}")

print("\n6. Sample of cleaned data:")
display(df_clean.head(10))

# Save the cleaned dataset
output_path = task1_dir / "task1_cleaned_data.csv"

df_clean.to_csv(output_path, index=False)

print("\nCleaned dataset saved to:")
print(output_path.resolve())
```
