```python
# ==================================================
# TASK 2: DATA TRANSFORMATION AND AGGREGATION
# ==================================================

import pandas as pd
from pathlib import Path

# Locate project directory
project_dir = Path.cwd()

if project_dir.name == "Task-2-Data-Transformation":
    project_dir = project_dir.parent

# Define input and output paths
input_file = (
    project_dir
    / "Task-1-Data-Exploration"
    / "task1_cleaned_data.csv"
)

output_dir = project_dir / "Task-2-Data-Transformation"

# Load cleaned dataset
df = pd.read_csv(input_file)

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

display(df.head())
```

```python
# ==================================================
# TASK 2.1: DATA FILTERING
# ==================================================

# Filter trains operating on Saturday
saturday_trains = df[
    df["days"].str.strip().str.lower() == "saturday"
].copy()

print("TASK 2.1 - SATURDAY TRAINS")
print("Total Saturday trains:", len(saturday_trains))

display(saturday_trains.head(10))

# Save filtered data
saturday_trains.to_csv(
    output_dir / "saturday_trains.csv",
    index=False
)

print("Saturday train data saved successfully!")
```

```python
# Filter trains starting from Chennai Beach
chennai_beach_trains = df[
    df["Source_Station_Name"].str.strip().str.upper()
    == "CHENNAI BEACH"
].copy()

print("TRAINS STARTING FROM CHENNAI BEACH")
print("Total trains:", len(chennai_beach_trains))

display(chennai_beach_trains.head(10))

# Save filtered data
chennai_beach_trains.to_csv(
    output_dir / "chennai_beach_trains.csv",
    index=False
)

print("Chennai Beach train data saved successfully!")
```

```python
# ==================================================
# TASK 2.2: GROUPING AND AGGREGATION
# ==================================================

# Count trains originating from each source station
trains_per_source_station = (
    df.groupby("Source_Station_Name")
    .size()
    .reset_index(name="Train_Count")
    .sort_values(by="Train_Count", ascending=False)
)

print("TRAIN COUNT BY SOURCE STATION")
display(trains_per_source_station.head(10))

# Save the result
trains_per_source_station.to_csv(
    output_dir / "trains_per_source_station.csv",
    index=False
)

print("Source station aggregation saved successfully!")
```

```python
# Calculate average trains per operating day for each source station

average_trains_per_day = (
    df.groupby("Source_Station_Name")
    .agg(
        Total_Trains=("Train_No", "count"),
        Operating_Days=("days", "nunique")
    )
    .reset_index()
)

average_trains_per_day["Average_Trains_Per_Day"] = (
    average_trains_per_day["Total_Trains"]
    / average_trains_per_day["Operating_Days"]
)

average_trains_per_day = average_trains_per_day.sort_values(
    by="Average_Trains_Per_Day",
    ascending=False
)

print("AVERAGE TRAINS PER OPERATING DAY BY SOURCE STATION")

display(average_trains_per_day.head(10))

# Save the result
average_trains_per_day.to_csv(
    output_dir / "average_trains_per_day.csv",
    index=False
)

print("Average train aggregation saved successfully!")
```

```python
# ==================================================
# TASK 2.3: DATA ENRICHMENT
# ==================================================

# Categorize trains based on operating days
df["Service_Type"] = df["days"].apply(
    lambda day: "Weekend"
    if day in ["Saturday", "Sunday"]
    else "Weekday"
)

print("SERVICE TYPE DISTRIBUTION")

display(df["Service_Type"].value_counts())

print("\nSAMPLE ENRICHED DATA")

display(df.head(10))

# Save enriched dataset
df.to_csv(
    output_dir / "railway_data_enriched.csv",
    index=False
)

print("Enriched railway dataset saved successfully!")
```
